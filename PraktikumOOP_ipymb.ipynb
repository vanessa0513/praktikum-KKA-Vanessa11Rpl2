Latihan 1: Class Hero

```python
# Definisi Class Hero
class Hero:
    # Constructor: Dijalankan otomatis saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
        self.name = name                 # Atribut Nama Hero
        self.hp = hp                     # Atribut Nyawa (Health Point)
        self.attack_power = attack_power # Atribut Kekuatan Serangan

    # Method untuk menampilkan status info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

# -- Main Program --

# 1. Membuat Objek (Instansiasi)
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# 2. Memanggil Method info() awal
hero1.info()
hero2.info()

# -- Tugas Analisis 1 --
# Mengubah nilai atribut hp pada hero1 secara langsung
hero1.hp = 500

print("\nSetelah HP Layla diubah:")
print(f"HP hero1 sekarang: {hero1.hp}")
hero1.info()
```

Latihan 2: Interaksi Antar Objek (Method)

```python
# Definisi Class Hero dengan Fitur Pertarungan
class Hero:
    # Constructor
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    # Method menampilkan status
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang: objek penyerang (self) menyerang objek target (lawan)
    def serang(self, lawan):
        print(f"\n {self.name} menyerang {lawan.name}!")
        # Memanggil method 'diserang' milik objek lawan dengan mengirimkan attack_power kita
        lawan.diserang(self.attack_power)

    # Method diserang: menerima damage dan mengurangi HP sendiri (self)
    def diserang(self, damage):
        self.hp -= damage
        print(f"  {self.name} terkena damage {damage}. Sisa HP: {self.hp}")


# -- Main Program --

# 1. Inisialisasi Objek Hero
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# 2. Status Awal
print("=== STATUS AWAL ===")
hero1.info()
hero2.info()

# 3. Simulasi Pertarungan (Interaksi Antar Objek)
print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2)  # Layla menyerang Zilong
hero2.serang(hero1)  # Zilong membalas menyerang Layla

# 4. Status Akhir
print("\n=== STATUS AKHIR ===")
hero1.info()
hero2.info()
```

Penjelasan Logika & Alur Interaksi
hero1.serang(hero2):

self merujuk ke Layla (hero1).

lawan merujuk ke objek utuh Zilong (hero2).

Layla memicu fungsi hero2.diserang(15) dengan mengirimkan nilai serangannya (self.attack_power).

lawan.diserang(...):

Di dalam method diserang, giliran Zilong yang menjadi self.

HP Zilong dikurangi sejumlah damage yang diterima (self.hp -= damage).

Nilai HP Zilong berkurang dari 120 menjadi 105.

Latihan 3: Pewarisan (Inheritance)

```python
# 1. Superclass / Parent Class
class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    def serang(self, lawan):
        print(f"\n {self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"  {self.name} terkena damage {damage}. Sisa HP: {self.hp}")


# 2. Subclass / Child Class: Mage (Mewarisi sifat dari Hero)
class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent Class (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana  # Atribut unik milik Mage

    # Method Overriding: Menimpa method info() dari Hero agar menampilkan Mana
    def info(self):
        print(f"\n {self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    # Method Unik milik Mage
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"\n {self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)  # Damage 2x lipat
        else:
            print(f"\n {self.name} gagal skill! Mana tidak cukup.")


# -- Main Program --
print("--- Update Class Hero (Inheritance) ---")

# Instansiasi Objek
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

# Cek Status Awal
eudora.info()
balmond.info()

# Pengujian Method
eudora.serang(balmond)         # Memakai method 'serang' (warisan dari Hero)
eudora.skill_fireball(balmond) # Memakai method khusus milik Mage

# Cek Status Akhir
eudora.info()
balmond.info()
```

Penjelasan Konsep OOP pada Latihan 3
Inheritance (class Mage(Hero)):

Class Mage secara otomatis mendapatkan semua atribut dan method dari class Hero tanpa perlu menulis ulang fungsi serang() atau diserang().

Fungsi super().__init__(...):

Menghubungkan inisialisasi class anak (Mage) ke class induk (Hero) agar atribut name, hp, dan attack_power diproses dan disimpan ke memori objek.

Method Overriding:

Mage mendefinisikan ulang method info(). Saat eudora.info() dipanggil, Python akan memprioritaskan fungsi info() versi Mage, bukan versi Hero.

Latihan 4: Enkapsulasi (Mengamankan Data HP)

```python
class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Tanda __ membuat atribut HP bersifat Private
        self.__hp = hp_awal

    # GETTER: Akses resmi untuk membaca nilai HP
    def get_hp(self):
        return self.__hp

    # SETTER: Akses resmi untuk mengubah HP dengan aturan/validasi
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0  # HP dicegat agar tidak boleh negatif
        elif nilai_baru > 1000:
            print(" Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        # Memakai getter dan setter di dalam class sendiri agar validasi aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f" {self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba --
hero1 = Hero("Layla", 100)

# Uji Coba Validasi Setter
hero1.set_hp(-50)
print(f"HP Hero1 setelah set_hp(-50) : {hero1.get_hp()}")

# Tugas Analisis 4: Uji Coba Name Mangling (Percobaan Hacking)
print(f"Mencoba akses paksa atribut private: {hero1._Hero__hp}")
```

4: Percobaan Hacking (hero1._Hero__hp):

Hasil Run: Nilai HP akan berhasil muncul (tidak menghasilkan error).

Penyebab (Name Mangling): Python mengubah nama variabel berawalan dua garis bawah (__hp) secara internal menjadi _NamaClass__namaVariabel (dalam hal ini _Hero__hp).

Alasan Etika Pemrograman: Enkapsulasi di Python berbasis kesepakatan komunitas (convention) dan prinsip kehati-hatian, bukan proteksi mutlak dari memori seperti Java/C++. Mengakses variabel dengan sintaks mangling melanggar abstraksi class dan memicu masalah sekunder jika nama class atau internal logika diubah di kemudian hari.

Uji Validasi (Menghapus Logika if-elif):

Hasil Run: Nilai HP menjadi -100.

Pentingnya Setter: Tanpa validasi di Setter, data atribut bisa bernilai tidak sah/anomali (seperti HP bernilai negatif atau overflow). Method Setter bertindak sebagai gerbang kendali (gatekeeper) yang memastikan integritas data tetap terjaga.

```python
from abc import ABC, abstractmethod

# 1. Abstract Class / Interface (Blueprint Kontrak)
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass  # Wajib diisi oleh class anak

    @abstractmethod
    def info(self):
        pass  # Wajib diisi oleh class anak

# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    def serang(self, target):
        print(f"🗡️ Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"🛡️ Saya adalah Hero: {self.nama}")

class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    def serang(self, target):
        print(f"👾 Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"👹 Saya adalah Monster: {self.jenis}")

# -- Uji Coba --
h = Hero("Alucard")
m = Monster("Serigala")

h.info()
m.info()
h.serang("Serigala")
m.serang("Alucard")
```

Penjelasan: Ketika sebuah class mewarisi Abstract Class / Interface, class anak tersebut wajib mengimplementasikan (override) seluruh method yang ditandai dengan @abstractmethod. Jika ada satu saja method abstrak yang terlewat, class anak tersebut otomatis dianggap sebagai class abstrak juga dan tidak bisa diinstansiasi menjadi objek.

Mencetak Cetakan (unit = GameUnit()):

Mengapa Dilarang: Class GameUnit bersifat abstrak/konseptual dan belum memiliki logika eksekusi pada method-methodnya (pass).

Fungsi Class Abstrak: Berperan sebagai kontrak kerangka kerja (blueprint standar) yang menjamin bahwa semua jenis unit game (Hero, Monster, NPC) di masa depan pasti memiliki antarmuka (interface) yang konsisten.

Latihan 6: Polymorphism (Fleksibilitas Interaksi)


```python
# 1. Superclass / Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")


# 2. Subclasses (Menerapkan Polymorphism dengan Method Overriding)
class Mage(Hero):
    def serang(self):
        print(f" {self.nama} (Mage) menembakkan Bola Api! Boom!")

class Archer(Hero):
    def serang(self):
        print(f" {self.nama} (Archer) memanah dari jauh! Jleb!")

class Fighter(Hero):
    def serang(self):
        print(f" {self.nama} (Fighter) memukul dengan pedang! Slash!")

# Menambahkan Class Baru: Healer (Tugas Analisis 1)
class Healer(Hero):
    def serang(self):
        print(f" {self.nama} (Healer) tidak menyerang, tapi menyembuhkan teman!")


# -- Penerapan Polymorphism --

# Daftar objek campuran dari berbagai class turunan
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord"),
    Healer("Estes")  # Objek Healer dimasukkan ke list pasukan
]

print("--- PERANG DIMULAI ---")
# Satu baris loop yang sama, mengeksekusi perilaku unik tiap objek
for pahlawan in pasukan:
    pahlawan.serang()
```

Penjelasan Konsep Polymorphism
Secara Bahasa: Poly (banyak) + Morph (bentuk).

Dalam Pemrograman: Kemampuan objek-objek dari berbagai class yang berbeda untuk merespons nama panggilan method yang sama dengan perilaku/logika internal yang unik.

Pada kode di atas, loop tidak perlu tahu apakah pahlawan saat itu bertipe Mage, Archer, atau Fighter. Loop cukup memanggil .serang(), dan Python secara otomatis memilih versi method milik class objek tersebut.

Keuntungan Polymorphism bagi Programmer saat Update Game:

Open/Closed Principle (Mudah Dikembangkan): Kode sistem utama (seperti loop battle, sistem giliran, atau antrean giliran) bersifat tertutup dari perubahan, namun terbuka untuk penambahan fitur baru.

Menghindari Percabangan Rumit (if-else / switch-case): Tanpa polimorfisme, programmer harus menulis kode pengecekan tipe manual. Dan dengan polimorfisme, kode di atas cukup digantikan satu baris: pahlawan.serang(). Saat game menambah 50 hero baru, loop pertarungan tidak perlu diubah sama sekali.

2. Uji Konsistensi Penamaan (tembak_panah)
Apa yang Terjadi saat dijalankan?

Objek Miya (Archer) tidak menjalankan tembak_panah, melainkan mencetak:

Ketika loop memanggil pahlawan.serang() pada giliran Miya, Python mencari method bernama serang di class Archer. (Catatan: Jika di class Hero juga tidak ada method serang, program akan langsung berhenti dengan error AttributeError).

Mengapa Nama Method Harus Persis Sama?

Polimorfisme bekerja berdasarkan kontrak antarmuka (interface contract). Nama method berfungsi sebagai tombol pengendali umum. Jika satu class anak mengubah nama fungsinya sendiri, sistem loop/kontroler pusat tidak akan bisa memicunya melalui jalur pemanggilan standar yang sama.
