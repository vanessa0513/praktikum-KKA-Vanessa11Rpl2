```python
from abc import ABC, abstractmethod

class KamarHotel(ABC):
    def __init__(self, nama_kamar, harga_dasar):
        self.nama_kamar = nama_kamar
        self.__harga_dasar = harga_dasar
        self.__stok = 0

    def get_harga_dasar(self):
        return self.__harga_dasar

    def get_stok(self):
        return self.__stok

    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(f"Gagal update stok {self.nama_kamar}! Stok tidak boleh negatif ({jumlah}).")
        else:
            self.__stok += jumlah
            print(f"Berhasil menambahkan stok {self.nama_kamar}: {self.__stok} unit.")

    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass
```

```python
class KamarDeluxe(KamarHotel):
    def __init__(self, harga_dasar, fasilitas, nama_tampil="Kamar Standard"):
        super().__init__("Kamar Deluxe", harga_dasar)
        self.fasilitas = fasilitas

    def tampilkan_detail(self):
        return f"[DELUXE] Kamar Deluxe Sea View | Fasilitas: {self.fasilitas}"

    def hitung_harga_total(self, jumlah_malam):
        harga_dasar = self.get_harga_dasar()
        pajak = 0.10 * harga_dasar
        subtotal = (harga_dasar + pajak) * jumlah_malam
        return harga_dasar, pajak, subtotal

class KamarStandard(KamarHotel):
    def __init__(self, harga_dasar, kapasitas, nama_tampil="Kamar Standard"):
        super().__init__("Kamar Standard", harga_dasar)
        self.kapasitas = kapasitas

    def tampilkan_detail(self):
        return f"[STANDARD] Kamar Standard Superior | Kapasitas: {self.kapasitas}"

    def hitung_harga_total(self, jumlah_malam):
        harga_dasar = self.get_harga_dasar()
        pajak = 0.05 * harga_dasar
        subtotal = (harga_dasar + pajak) * jumlah_malam
        return harga_dasar, pajak, subtotal
```

```python
def proses_transaksi(daftar_pesanan):
    print("\n--- STRUK PEMESANAN ---")
    total_tagihan = 0

    for i, (kamar, jumlah_malam) in enumerate(daftar_pesanan, 1):
        detail = kamar.tampilkan_detail()
        harga_dasar, pajak, subtotal = kamar.hitung_harga_total(jumlah_malam)

        persen_pajak = "10%" if isinstance(kamar, KamarDeluxe) else "5%"

        print(f"{i}. {detail}")
        print(f"Harga Dasar/Malam: Rp {harga_dasar:,.0f} | Pajak({persen_pajak}): Rp {pajak:,.0f}".replace(",", "."))
        print(f"Menginap: {jumlah_malam} malam | Subtotal: Rp {subtotal:,.0f}".replace(",", "."))

        total_tagihan += subtotal

    print("-" * 40)
    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}".replace(",", "."))
    print("-" * 40)

# Menjalankan skenario
print("--- SETUP DATA KAMAR ---")

kamar_deluxe = KamarDeluxe(1500000, "Private Pool")
kamar_deluxe.tambah_stok(10)

kamar_standard = KamarStandard(500000, "2 Orang")
kamar_standard.tambah_stok(-5)
kamar_standard.tambah_stok(20)

pesanan_tamu = [
    (kamar_deluxe, 2),
    (kamar_standard, 1)
]

proses_transaksi(pesanan_tamu)
```
